```python
from openai import OpenAI

# initialize client – will use the key from environment variable OPENAI_API_KEY
client = OpenAI()

try:
    # lightweight test call
    response = client.models.list()
    print("API key is valid. Available models:")
    for m in response.data[:5]:
        print(" -", m.id)

except Exception as e:
    print("API key test failed:")
    print(e)
```

```python
# test_env_key.py
from dotenv import load_dotenv
from openai import OpenAI
import os

# load variables from .env in current directory
load_dotenv()
key = os.getenv("OPENAI_API_KEY")

if not key:
    print("No key found")
    exit()

client = OpenAI(api_key=key)

try:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": "test"}],
        max_tokens=5
    )

    print("API key valid")
    print("Response:", response.choices[0].message.content)

except Exception as e:
    print("Request failed")
    print(e)
```
