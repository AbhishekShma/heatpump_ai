```python
import pandas as pd
import numpy as np

def calculate_design_temperature(series, reference_years=20, n_lowest=10):
    """
    Calculate design temperature using DIN-like approach.
    
    Args:
        series: pandas Series with year as index and temperature values
        reference_years: Number of years for reference period (default 20)
        n_lowest: Number of lowest values to consider (default 10)
    
    Returns:
        Design temperature value
    """
    # Calculate 2-year rolling averages
    rolling_avg = series.rolling(window=2, min_periods=2).mean()
    
    # Use the most recent reference_years period
    recent_data = rolling_avg.tail(reference_years)
    
    # Remove NaN values (from first year which can't have 2-year average)
    recent_data = recent_data.dropna()
    
    # Find the n_lowest 2-year averages
    lowest_values = recent_data.nsmallest(n_lowest)
    
    # Design temperature is the nth lowest (i.e., the maximum of the n lowest)
    design_temp = lowest_values.max()
    
    return design_temp

# Read the cleaned data
df = pd.read_csv("cleaned_aggregated_yearly_data.csv")

# Set Jahr as index for easier processing
df_indexed = df.set_index('Jahr')

# Calculate design temperature for each column
design_temps = {}
for col in df_indexed.columns:
    design_temps[col] = calculate_design_temperature(df_indexed[col])

# Create a DataFrame with one row containing design temperatures
design_df = pd.DataFrame([design_temps])

# Save to CSV
design_df.to_csv("design_temperatures.csv", index=False)
print("Design temperatures calculated and saved to design_temperatures.csv")
print("\nDesign Temperatures:")
print(design_df)
```

# Convert allCountries.txt (Tab-Separated) to CSV


```python
def convert_tsv_to_csv(input_file, output_file, chunk_size=10000):
    """
    Convert a large tab-separated file to CSV format.
    
    Processes the file in chunks to handle large files efficiently.
    
    Args:
        input_file: Path to input tab-separated file
        output_file: Path to output CSV file
        chunk_size: Number of lines to process at a time (default: 10000)
    """
    import csv
    
    print(f"Converting {input_file} to {output_file}...")
    
    with open(input_file, 'r', encoding='utf-8') as infile, \
         open(output_file, 'w', encoding='utf-8', newline='') as outfile:
        
        # Create CSV writer
        csv_writer = csv.writer(outfile)
        
        lines_processed = 0
        chunk = []
        
        for line in infile:
            # Strip whitespace and split by tab
            row = line.rstrip('\n\r').split('\t')
            chunk.append(row)
            
            # Write chunk when it reaches chunk_size
            if len(chunk) >= chunk_size:
                csv_writer.writerows(chunk)
                lines_processed += len(chunk)
                chunk = []
                if lines_processed % 100000 == 0:
                    print(f"  Processed {lines_processed:,} lines...")
        
        # Write remaining lines
        if chunk:
            csv_writer.writerows(chunk)
            lines_processed += len(chunk)
    
    print(f"Conversion complete! Processed {lines_processed:,} lines.")
    print(f"Output saved to: {output_file}")

# Convert the file
input_file = "allCountries.txt"
output_file = "allCountries.csv"

convert_tsv_to_csv(input_file, output_file)
```

# Extract German locations (Country Code: DE) from allCountries.csv


```python
def extract_country_rows(input_file, output_file, country_code, chunk_size=50000, add_header=True):
    """
    Extract rows with a specific country code from a large CSV file.
    
    Processes the file in chunks to handle large files efficiently.
    
    Args:
        input_file: Path to input CSV file
        output_file: Path to output CSV file
        country_code: Country code to filter (e.g., 'DE' for Germany)
        chunk_size: Number of lines to process at a time (default: 50000)
        add_header: Whether to add a header row to the output (default: True)
    """
    import csv
    
    print(f"Extracting rows with country code '{country_code}' from {input_file}...")
    
    rows_extracted = 0
    
    with open(input_file, 'r', encoding='utf-8') as infile, \
         open(output_file, 'w', encoding='utf-8', newline='') as outfile:
        
        csv_reader = csv.reader(infile)
        csv_writer = csv.writer(outfile)
        
        # Add header if requested
        if add_header:
            header = ['Country Code', 'Postal Code', 'Place Name', 'State/Province', 
                     'State Code', 'County', 'County Code', 'Community', 'Community Code',
                     'Latitude', 'Longitude', 'Accuracy']
            csv_writer.writerow(header)
        
        chunk = []
        lines_processed = 0
        
        for row in csv_reader:
            # Check if first column matches country code
            if row and row[0] == country_code:
                chunk.append(row)
                rows_extracted += 1
            
            lines_processed += 1
            
            # Write chunk when it reaches chunk_size
            if len(chunk) >= chunk_size:
                csv_writer.writerows(chunk)
                chunk = []
                if rows_extracted % 10000 == 0 and rows_extracted > 0:
                    print(f"  Extracted {rows_extracted:,} rows (processed {lines_processed:,} lines)...")
        
        # Write remaining rows
        if chunk:
            csv_writer.writerows(chunk)
    
    print(f"Extraction complete!")
    print(f"  Total rows extracted: {rows_extracted:,}")
    print(f"  Total lines processed: {lines_processed:,}")
    print(f"  Output saved to: {output_file}")

# Extract German locations
input_file = "allCountries.csv"
output_file = "allCountries_DE.csv"
country_code = "DE"

extract_country_rows(input_file, output_file, country_code)
```

# Extract Postal Code and State mapping from allCountries_DE.csv


```python
def extract_postal_code_state(input_file, output_file):
    """
    Extract postal code and state columns from the German locations CSV.
    
    Creates a CSV file with postal code and state mapping, removing duplicates.
    
    Args:
        input_file: Path to input CSV file (allCountries_DE.csv)
        output_file: Path to output CSV file
    """
    import csv
    
    print(f"Extracting postal code and state from {input_file}...")
    
    postal_state_map = {}
    rows_processed = 0
    
    with open(input_file, 'r', encoding='utf-8') as infile:
        csv_reader = csv.reader(infile)
        
        # Skip header
        header = next(csv_reader)
        postal_idx = 1  # Postal Code is 2nd column (index 1)
        state_idx = 3   # State/Province is 4th column (index 3)
        
        for row in csv_reader:
            if len(row) > max(postal_idx, state_idx):
                postal_code = row[postal_idx].strip()
                state = row[state_idx].strip()
                
                # Store mapping (postal code -> state)
                # If same postal code appears with different states, keep the first one
                if postal_code and state and postal_code not in postal_state_map:
                    postal_state_map[postal_code] = state
            
            rows_processed += 1
            if rows_processed % 10000 == 0:
                print(f"  Processed {rows_processed:,} rows...")
    
    # Write to output file
    with open(output_file, 'w', encoding='utf-8', newline='') as outfile:
        csv_writer = csv.writer(outfile)
        
        # Write header
        csv_writer.writerow(['Postal Code', 'State'])
        
        # Write data sorted by postal code
        for postal_code in sorted(postal_state_map.keys()):
            csv_writer.writerow([postal_code, postal_state_map[postal_code]])
    
    print(f"Extraction complete!")
    print(f"  Total rows processed: {rows_processed:,}")
    print(f"  Unique postal codes: {len(postal_state_map):,}")
    print(f"  Output saved to: {output_file}")

# Extract postal code and state
input_file = "allCountries_DE.csv"
output_file = "postal_code_state_DE.csv"

extract_postal_code_state(input_file, output_file)
```

# Map Postal Codes to Design Temperatures


```python
def map_postal_code_to_temperature(postal_state_file, design_temp_file, output_file):
    """
    Map postal codes to design temperatures based on their state/region.
    
    Args:
        postal_state_file: Path to postal_code_state_DE.csv
        design_temp_file: Path to design_temperatures.csv
        output_file: Path to output CSV file
    """
    import csv
    
    # Mapping from state names in postal code file to region names in design temperatures
    state_to_region_map = {
        'Baden-Württemberg': 'Baden-Wuerttemberg',
        'Bayern': 'Bayern',
        'Bavaria': 'Bayern',
        'Hessen': 'Hessen',
        'Niedersachsen': 'Niedersachsen',
        'Lower Saxony': 'Niedersachsen',
        'Nordrhein-Westfalen': 'Nordrhein-Westfalen',
        'Rheinland-Pfalz': 'Rheinland-Pfalz',
        'Schleswig-Holstein': 'Schleswig-Holstein',
        'Saarland': 'Saarland',
        'Sachsen': 'Sachsen',
        'Saxony': 'Sachsen',
        'Sachsen-Anhalt': 'Sachsen-Anhalt',
        'Saxony-Anhalt': 'Sachsen-Anhalt',
        'Thueringen': 'Thueringen',
        'Thüringen': 'Thueringen',
        'Thuringia': 'Thueringen',
        'Mecklenburg-Vorpommern': 'Mecklenburg-Vorpommern',
        'Mecklenburg-Western Pomerania': 'Mecklenburg-Vorpommern',
        'Brandenburg': 'Brandenburg',
        'Berlin': 'Brandenburg/Berlin',
        'Land Berlin': 'Brandenburg/Berlin',
        'Hamburg': 'Niedersachsen/Hamburg/Bremen',
        'Bremen': 'Niedersachsen/Hamburg/Bremen',
    }
    
    print(f"Loading design temperatures from {design_temp_file}...")
    
    # Load design temperatures
    with open(design_temp_file, 'r', encoding='utf-8') as f:
        reader = csv.reader(f)
        regions = next(reader)  # Region names
        temp_values = next(reader)  # Temperature values
    
    # Create region to temperature mapping
    region_temp_map = {}
    for region, temp in zip(regions, temp_values):
        region_temp_map[region] = float(temp)
    
    print(f"Loaded temperatures for {len(region_temp_map)} regions")
    print(f"\nRegion temperatures:")
    for region, temp in sorted(region_temp_map.items()):
        if region != 'Deutschland':  # Skip overall Germany average
            print(f"  {region}: {temp}°C")
    
    print(f"\nProcessing postal codes from {postal_state_file}...")
    
    # Process postal codes and map to temperatures
    postal_temp_data = []
    unmapped_states = set()
    rows_processed = 0
    
    with open(postal_state_file, 'r', encoding='utf-8') as f:
        reader = csv.reader(f)
        header = next(reader)  # Skip header
        
        for row in reader:
            if len(row) >= 2:
                postal_code = row[0].strip()
                state = row[1].strip()
                
                # Map state to region
                region = state_to_region_map.get(state)
                
                if region and region in region_temp_map:
                    temperature = region_temp_map[region]
                    postal_temp_data.append([postal_code, state, region, temperature])
                else:
                    unmapped_states.add(state)
                    # Try to find a partial match
                    found = False
                    for state_key, region_key in state_to_region_map.items():
                        if state_key.lower() in state.lower() or state.lower() in state_key.lower():
                            if region_key in region_temp_map:
                                temperature = region_temp_map[region_key]
                                postal_temp_data.append([postal_code, state, region_key, temperature])
                                found = True
                                break
                    if not found:
                        # Use Germany average as fallback
                        if 'Deutschland' in region_temp_map:
                            temperature = region_temp_map['Deutschland']
                            postal_temp_data.append([postal_code, state, 'Deutschland', temperature])
            
            rows_processed += 1
            if rows_processed % 10000 == 0:
                print(f"  Processed {rows_processed:,} rows...")
    
    # Write output file
    print(f"\nWriting output to {output_file}...")
    with open(output_file, 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['Postal Code', 'State', 'Region', 'Design Temperature (°C)'])
        
        # Sort by postal code
        for row in sorted(postal_temp_data, key=lambda x: x[0]):
            writer.writerow(row)
    
    print(f"\nMapping complete!")
    print(f"  Total postal codes processed: {rows_processed:,}")
    print(f"  Postal codes with temperature mapping: {len(postal_temp_data):,}")
    if unmapped_states:
        print(f"  Warning: Found {len(unmapped_states)} unmapped state(s): {sorted(unmapped_states)}")
    print(f"  Output saved to: {output_file}")

# Map postal codes to temperatures
postal_state_file = "postal_code_state_DE.csv"
design_temp_file = "design_temperatures.csv"
output_file = "postal_code_temperature_DE.csv"

map_postal_code_to_temperature(postal_state_file, design_temp_file, output_file)
```

# Map Postal Codes to Design Temperatures (using design_temperatures_2.csv)


```python
def map_postal_code_to_temperature_v2(postal_state_file, design_temp_file, output_file):
    """
    Map postal codes to design temperatures based on their state/region.
    Uses design_temperatures_2.csv which has separate entries for Berlin, Bremen, Hamburg.
    
    Args:
        postal_state_file: Path to postal_code_state_DE.csv
        design_temp_file: Path to design_temperatures_2.csv
        output_file: Path to output CSV file
    """
    import csv
    
    # Mapping from state names in postal code file to region names in design_temperatures_2.csv
    # Note: design_temperatures_2.csv has separate entries for Berlin, Bremen, Hamburg
    state_to_region_map = {
        'Baden-Württemberg': 'Baden-Württemberg',
        'Bayern': 'Bayern',
        'Bavaria': 'Bayern',
        'Hessen': 'Hessen',
        'Niedersachsen': 'Niedersachsen',
        'Lower Saxony': 'Niedersachsen',
        'Nordrhein-Westfalen': 'Nordrhein-Westfalen',
        'Rheinland-Pfalz': 'Rheinland-Pfalz',
        'Schleswig-Holstein': 'Schleswig-Holstein',
        'Saarland': 'Saarland',
        'Sachsen': 'Sachsen',
        'Saxony': 'Sachsen',
        'Sachsen-Anhalt': 'Sachsen-Anhalt',
        'Saxony-Anhalt': 'Sachsen-Anhalt',
        'Thueringen': 'Thüringen',
        'Thüringen': 'Thüringen',
        'Thuringia': 'Thüringen',
        'Mecklenburg-Vorpommern': 'Mecklenburg-Vorpommern',
        'Mecklenburg-Western Pomerania': 'Mecklenburg-Vorpommern',
        'Brandenburg': 'Brandenburg',
        'Berlin': 'Berlin',
        'Land Berlin': 'Berlin',
        'Hamburg': 'Hamburg',
        'Bremen': 'Bremen',
    }
    
    print(f"Loading design temperatures from {design_temp_file}...")
    
    # Load design temperatures (try different encodings)
    encodings = ['utf-8', 'latin-1', 'cp1252', 'iso-8859-1']
    regions = None
    temp_values = None
    
    for encoding in encodings:
        try:
            with open(design_temp_file, 'r', encoding=encoding) as f:
                reader = csv.reader(f)
                regions = next(reader)  # Region names
                temp_values = next(reader)  # Temperature values
            print(f"Successfully loaded file with {encoding} encoding")
            break
        except (UnicodeDecodeError, UnicodeError):
            continue
    
    if regions is None:
        raise ValueError("Could not read file with any encoding")
    
    # Filter out empty columns
    region_temp_map = {}
    for i, region in enumerate(regions):
        if region and region.strip() and i < len(temp_values) and temp_values[i]:
            region_temp_map[region.strip()] = float(temp_values[i])
    
    print(f"Loaded temperatures for {len(region_temp_map)} regions")
    print(f"\nRegion temperatures:")
    for region, temp in sorted(region_temp_map.items()):
        print(f"  {region}: {temp}°C")
    
    print(f"\nProcessing postal codes from {postal_state_file}...")
    
    # Process postal codes and map to temperatures
    postal_temp_data = []
    unmapped_states = set()
    rows_processed = 0
    
    with open(postal_state_file, 'r', encoding='utf-8') as f:
        reader = csv.reader(f)
        header = next(reader)  # Skip header
        
        for row in reader:
            if len(row) >= 2:
                postal_code = row[0].strip()
                state = row[1].strip()
                
                # Map state to region
                region = state_to_region_map.get(state)
                
                if region and region in region_temp_map:
                    temperature = region_temp_map[region]
                    postal_temp_data.append([postal_code, state, region, temperature])
                else:
                    unmapped_states.add(state)
                    # Try to find a partial match
                    found = False
                    for state_key, region_key in state_to_region_map.items():
                        if state_key.lower() in state.lower() or state.lower() in state_key.lower():
                            if region_key in region_temp_map:
                                temperature = region_temp_map[region_key]
                                postal_temp_data.append([postal_code, state, region_key, temperature])
                                found = True
                                break
                    if not found:
                        # Try direct match with any region name
                        for region_name in region_temp_map.keys():
                            if state.lower() in region_name.lower() or region_name.lower() in state.lower():
                                temperature = region_temp_map[region_name]
                                postal_temp_data.append([postal_code, state, region_name, temperature])
                                found = True
                                break
                    if not found:
                        # Use average of all regions as fallback
                        if region_temp_map:
                            avg_temp = sum(region_temp_map.values()) / len(region_temp_map)
                            postal_temp_data.append([postal_code, state, 'Average', avg_temp])
            
            rows_processed += 1
            if rows_processed % 10000 == 0:
                print(f"  Processed {rows_processed:,} rows...")
    
    # Write output file
    print(f"\nWriting output to {output_file}...")
    with open(output_file, 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['Postal Code', 'State', 'Region', 'Design Temperature (°C)'])
        
        # Sort by postal code
        for row in sorted(postal_temp_data, key=lambda x: x[0]):
            writer.writerow(row)
    
    print(f"\nMapping complete!")
    print(f"  Total postal codes processed: {rows_processed:,}")
    print(f"  Postal codes with temperature mapping: {len(postal_temp_data):,}")
    if unmapped_states:
        print(f"  Warning: Found {len(unmapped_states)} unmapped state(s): {sorted(unmapped_states)}")
    print(f"  Output saved to: {output_file}")

# Map postal codes to temperatures using design_temperatures_2.csv
postal_state_file = "postal_code_state_DE.csv"
design_temp_file = "design_temperatures_2.csv"
output_file = "postal_code_temperature_DE_2.csv"

map_postal_code_to_temperature_v2(postal_state_file, design_temp_file, output_file)
```
