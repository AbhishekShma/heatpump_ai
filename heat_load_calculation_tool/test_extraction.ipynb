# Test JSON Extraction Tool

This notebook tests the JSON extraction tool with a sample building conversation summary.

```python
import json
from json_extraction_tool import extract_and_store_json
```

## Sample Summary

Sample conversation summary containing building information:

```python
# Sample summary matching the expected format
SAMPLE_SUMMARY = """The building is a two-story house located in Munich with postal code 81248. 
It was constructed in 1995. The heated floor area is 100 square meters per floor. 
The building has two walls that are touching adjacent buildings. 

The house underwent renovations in 2015, specifically the windows were replaced that year. 
The roof was also renovated and insulated. However, the walls and floor were not renovated. 
The walls are not insulated.

Each floor has a height of 2.5 meters. The indoor design temperature is set to 21 degrees Celsius. 
The building has both a ground floor and a top floor."""

print("Input Summary:")
print("-" * 50)
print(SAMPLE_SUMMARY)
print("-" * 50)
```

## Run Extraction

```python
result = extract_and_store_json(SAMPLE_SUMMARY, output_dir="test_outputs")

print("Extraction Result:")
print("-" * 50)
print(f"Success: {result['success']}")
print("-" * 50)
```

## Display Results

```python
if result['success']:
    print("Extracted JSON:")
    print(json.dumps(result['json_data'], indent=2))
    print(f"\nSaved to: {result['file_path']}")
else:
    print(f"Error: {result['error']}")
```
