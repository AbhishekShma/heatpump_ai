```python
import pandas as pd

INPUT_CSV = "/mnt/d/heatpump_data/heatpump_table/consolidated_table_1.csv"
OUTPUT_CSV = "/mnt/d/heatpump_data/heatpump_table/consolidated_table_1_numeric_values_corrected.csv"

# Columns that contain European-formatted numbers
NUMERIC_COLUMNS = [
    "heat_output_35_C_in_kw",
    "effeciency_35_C",
    "heat_output_55_C_in_kw",
    "effeciency_55_C",
]

def normalize_eu_number(x):
    if pd.isna(x):
        return x
    x = str(x).strip()
    if x == "":
        return x
    return x.replace(".", "").replace(",", ".")

# Load everything as text
df = pd.read_csv(INPUT_CSV)
# df
# Apply transformation ONLY to numeric columns
for col in NUMERIC_COLUMNS:
    if col in df.columns:
        df[col] = df[col].apply(normalize_eu_number)

df
# Write cleaned CSV
df.to_csv(OUTPUT_CSV, index=False)
```
