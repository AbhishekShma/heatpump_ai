```python
CSV_PATH = "/mnt/d/heatpump_data/heatpump_table/consolidated_table_1_numeric_values_corrected.csv"
```

```python
import csv
import psycopg2
from decimal import Decimal, InvalidOperation
from datetime import datetime
from tqdm.auto import tqdm


# ---------- helpers ----------

def parse_decimal(value):
    """
    Safely parse numeric values.
    Returns Decimal or None.
    """
    if value is None:
        return None

    value = str(value).strip()
    if value == "":
        return None

    try:
        return Decimal(value)
    except InvalidOperation:
        return None


# ---------- main ----------

conn = psycopg2.connect(DATABASE_URL)
conn.autocommit = False

BATCH_SIZE = 1000
now = datetime.now()

try:
    with conn.cursor() as cur:

        # Drop & recreate table
        cur.execute("DROP TABLE IF EXISTS heat_pump_models")

        cur.execute("""
            CREATE TABLE heat_pump_models (
                id SERIAL PRIMARY KEY,
                manufacturer VARCHAR(150) NOT NULL,
                model_type VARCHAR(200),
                heat_output_35_kw NUMERIC,
                cop_35 NUMERIC,
                heat_output_55_kw NUMERIC,
                cop_55 NUMERIC,
                refrigerant VARCHAR(20),
                grid_service_availability VARCHAR(20),
                ee_display_availability VARCHAR(20),
                category VARCHAR(150),
                created_at TIMESTAMPTZ NOT NULL,
                updated_at TIMESTAMPTZ NOT NULL
            )
        """)

        with open(CSV_PATH, newline="", encoding="utf-8") as f:
            reader = csv.DictReader(f)
            rows = list(reader)

            for i, row in enumerate(
                tqdm(rows, desc="Uploading heat pump data to DB"), start=1
            ):
                cur.execute("""
                    INSERT INTO heat_pump_models (
                        manufacturer,
                        model_type,
                        heat_output_35_kw,
                        cop_35,
                        heat_output_55_kw,
                        cop_55,
                        refrigerant,
                        grid_service_availability,
                        ee_display_availability,
                        category,
                        created_at,
                        updated_at
                    )
                    VALUES (
                        %(manufacturer)s,
                        %(model_type)s,
                        %(heat_output_35_kw)s,
                        %(cop_35)s,
                        %(heat_output_55_kw)s,
                        %(cop_55)s,
                        %(refrigerant)s,
                        %(grid_service_availability)s,
                        %(ee_display_availability)s,
                        %(category)s,
                        %(created_at)s,
                        %(updated_at)s
                    )
                """, {
                    "manufacturer": row["Manufacturer"],
                    "model_type": row["Type"],
                    "heat_output_35_kw": parse_decimal(row["heat_output_35_C_in_kw"]),
                    "cop_35": parse_decimal(row["effeciency_35_C"]),
                    "heat_output_55_kw": parse_decimal(row["heat_output_55_C_in_kw"]),
                    "cop_55": parse_decimal(row["effeciency_55_C"]),
                    "refrigerant": row["Refrigerant"],
                    "grid_service_availability": row["grid_service_availability"],
                    "ee_display_availability": row["ee_display_availibility"],
                    "category": row["category"],
                    "created_at": now,
                    "updated_at": now,
                })

                # Commit in batches
                if i % BATCH_SIZE == 0:
                    conn.commit()

        # Final commit
        conn.commit()

finally:
    conn.close()
```
