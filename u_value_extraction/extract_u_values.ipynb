# Extract U-Values from Tabula.xlsm

This notebook extracts all columns for Germany (DE) from the "Tab.Building.Constr" sheet in the Tabula.xlsm file.

```python
import pandas as pd
import os
from pathlib import Path
```

```python
# Path to the Excel file
excel_path = "/mnt/d/heatpump_data/TABULA_xlsm_Download/Tabula.xlsm"
sheet_name = "Tab.Building.Constr"

# Verify file exists
if not os.path.exists(excel_path):
    raise FileNotFoundError(f"Excel file not found at: {excel_path}")

print(f"Reading Excel file: {excel_path}")
print(f"Target sheet: {sheet_name}")
```

```python
# Load the Excel file and list all sheet names
xl_file = pd.ExcelFile(excel_path, engine='openpyxl')
print("Available sheets:")
for idx, sheet in enumerate(xl_file.sheet_names, 1):
    print(f"  {idx}. {sheet}")
    
# Verify the target sheet exists
if sheet_name not in xl_file.sheet_names:
    raise ValueError(f"Sheet '{sheet_name}' not found in the Excel file!")
```

```python
# Read the target sheet
df = pd.read_excel(excel_path, sheet_name=sheet_name, engine='openpyxl')

print(f"Shape of the data: {df.shape}")
print(f"\nColumn names:")
print(df.columns.tolist())
print(f"\nFirst few rows:")
df.head()
```

```python
# Check for country code column (likely named something like 'Country', 'Country Code', 'ISO', 'DE', etc.)
print("Checking for country-related columns:")
country_cols = [col for col in df.columns if any(keyword in str(col).lower() for keyword in ['country', 'iso', 'code', 'nation'])]
print(f"Potential country columns: {country_cols}")

# Display unique values in potential country columns
for col in country_cols[:3]:  # Check first 3 potential columns
    print(f"\nUnique values in '{col}':")
    print(df[col].unique()[:20])  # Show first 20 unique values
```

```python
# Filter for Germany (DE) - adjust column name and value based on actual data structure
# Common patterns: 'DE', 'Germany', 'GER', etc.

# Try to find the country column automatically
country_column = None
for col in df.columns:
    unique_vals = df[col].dropna().astype(str).unique()
    if any(val.upper() in ['DE', 'GERMANY', 'GER'] for val in unique_vals):
        country_column = col
        print(f"Found country column: '{col}'")
        print(f"Unique values: {df[col].dropna().unique()[:10]}")
        break

if country_column is None:
    print("Could not automatically detect country column. Please inspect the data above.")
    print("You may need to manually specify the country column name.")
else:
    # Filter for DE/Germany
    de_mask = df[country_column].astype(str).str.upper().isin(['DE', 'GERMANY', 'GER'])
    df_de = df[de_mask].copy()
    
    print(f"\nFiltered data for Germany:")
    print(f"  Original rows: {len(df)}")
    print(f"  Germany rows: {len(df_de)}")
    print(f"  Columns: {len(df_de.columns)}")
```

```python
# Display the filtered data
if 'df_de' in locals():
    print("Sample of Germany data:")
    df_de.head(10)
else:
    print("Please run the previous cell first to filter the data.")
```

```python
# Save the extracted data
if 'df_de' in locals():
    output_dir = Path("output")
    output_dir.mkdir(exist_ok=True)
    
    # Save as CSV
    csv_path = output_dir / "germany_u_values.csv"
    df_de.to_csv(csv_path, index=False)
    print(f"Saved CSV to: {csv_path}")
    
    # Save as Excel
    excel_output_path = output_dir / "germany_u_values.xlsx"
    df_de.to_excel(excel_output_path, index=False, engine='openpyxl')
    print(f"Saved Excel to: {excel_output_path}")
    
    print(f"\nSummary:")
    print(f"  Total rows: {len(df_de)}")
    print(f"  Total columns: {len(df_de.columns)}")
else:
    print("No data to save. Please run the filtering cell first.")
```
